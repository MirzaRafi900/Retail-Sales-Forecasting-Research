# Visualization

## Sales Distribution Analysis

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Sales"],
    bins=50,
    color="orange"
)

plt.title("Sales Distribution Analysis")

plt.xlabel("Sales")

plt.ylabel("Frequency")

plt.savefig(
    "../Figures/Sales_Distributiom_Analysis.jpeg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Observation

Sales distribution is positively skewed.

Most observations are concentrated at lower
sales values, while a smaller number exhibit
significantly higher sales volumes.

This indicates the presence of sales spikes.
and variability in retail demand.

## Store Type VS Sales

```python
df.groupby("StoreType")["Sales"].mean()

sns.boxplot(
    x="StoreType",
    y="Sales",
    data=df
)

plt.savefig(
    "../Figures/StoreTypes_VS_Sales.jpeg",
    dpi=300,
    bbox_inches="tight"
)
```

### Observation

The sales distribution differs across store types.

Store Type B exhibits a noticeably higher median sales value.
and greater variability compared to the other store types.
Store types A, C, and D show relatively similar sales.
distributions.

This suggests that store characteristics may significantly
influence sales performance and should be considered during
the forecasting model development process.

```python
df.groupby("StoreType")["Sales"].mean().sort_values(
    ascending=False
)

sns.boxplot(
    x="StoreType",
    y="Sales",
    data=df
)

plt.savefig(
    "../Figures/StoreType_Vs_Sales_Average.jpeg",
    dpi=300,
    bbox_inches="tight"
)
```

```python
df.groupby("StoreType")["Sales"].agg(
    ["mean","median","min","max"]
).sort_values(
    by="mean",
    ascending=False
)
```

### Observation

Analysis of sales performance across store types
revealed notable differences.

Store Type B generated the highest average sales.
(10,058.84) and median sales (9,025.5), which are
substantially greater than those observed for
Store Types A, C, and D.

Store Types A, C, and D exhibited relatively
similar sales characteristics, with mean sales
around 5,700.

These findings suggest that store type is an
important explanatory variable and may contribute
significantly to forecasting accuracy.

## Promotion VS Sales

```python
df.groupby("Promo")["Sales"].agg(
    ["mean","median"]
)

sns.boxplot(
    x="Promo",
    y="Sales",
    data=df
)

plt.savefig(
    "../Figures/Promo_VS_Sales.jpeg",
    dpi=300,
    bbox_inches="tight"
)

```

```python
df.groupby("Promo")["Sales"].agg(
    ["mean","median","min","max"]
)
```

### Observation

Promotion activities appear to have a substantial
positive impact on sales performance.

Stores without promotions recorded an average
sales value of 4,406.05 and a median sales value
of 4,622.

In contrast, stores operating promotional campaigns
achieved an average sales value of 7,991.15 and
a median sales value of 7,553.

This represents an increase of approximately 81%
in average sales during promotional periods.

The boxplot further supports this finding, showing
consistently higher sales distributions when
promotions are active.

These results suggest that promotional activities
are a highly influential predictor of retail sales
and should be included in forecasting models.

## Weekly Sales

```python
df.groupby("DayOfWeek")["Sales"].agg(
    ["mean","median","min","max"]
)
```

```python
df.groupby("DayOfWeek")["Sales"].agg(
    ["mean","median","min","max"]
)

sns.boxplot(
    x="DayOfWeek",
    y="Sales",
    data=df
)

plt.savefig(
    "../Figures/Weekly_Sales.jpeg",
    dpi=300,
    bbox_inches="tight"
)
```

### Observation

Sales performance varies across different days of the week.

Day 1 recorded the highest average sales value
(7,809.04), while Day 7 recorded the lowest
average sales value (204.18).

The median sales value for Day 7 was zero,
suggesting that many stores generated no sales
on this day.

These findings indicate that the day of the week
has a substantial influence on retail sales and
should be included as a predictor variable in
forecasting models.

## Open Sales 

```python
df.groupby("Open")["Sales"].agg(
    ["mean","median","min","max"]
)
```

```python
df.groupby("Open")["Sales"].agg(
    ["mean","median","min","max"]
)

sns.boxplot(
    x="Open",
    y="Sales",
    data=df
)

plt.savefig(
    "../Figures/Open_Sales.jpeg",
    dpi=300,
    bbox_inches="tight"
)
```

## Customer VS Sales

```python
df[["Customers","Sales"]].corr()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x="Customers",
    y="Sales",
    data=df.sample(5000)
)
plt.title("Customer VS Sales")

plt.xlabel("Customer")
plt.ylabel("Sales")

plt.savefig(
    "../Figures/Customer_VS_Sales.jpeg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Observation

A very strong positive relationship exists between
customer count and sales performance.

Correlation analysis produced a coefficient of 0.89
between Customers and Sales, indicating a strong
linear association.

The scatter plot further confirms this relationship,
showing that sales generally increase as customer
traffic increases.

This finding suggests that customer count is one of
the strongest predictors of retail sales and is
expected to contribute significantly to forecasting
accuracy.

## Correlation Matrix

```python
numeric_df=df.select_dtypes(
    include=["int64","float64"]
)

plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    annot=True
)

plt.title("Correlation Matrix")

plt.savefig(
    "../Figures/Correlation_Matrix.jpeg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Observation

Correlation analysis identified customers,
Open status, promo status, and day of week as the
most influential variables associated with sales.

Customers demonstrated the strongest positive
correlation with sales (0.89), followed by open
status (0.68) and promo status (0.45).

DayOfWeek exhibited a moderate negative
correlation (-0.46), suggesting that sales
Patterns vary throughout the week.

These variables will be prioritized during
feature selection and forecasting model
development.
