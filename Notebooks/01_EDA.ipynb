# Exploratory Data Analysis (EDA)

## Missing Values

```python
df.isnull().sum()
```

## Observation

The dataset contains missing values primarily in
competition-related and promotion-related variables.

Missing values were observed in:

- CompetitionDistance (2,642)
- CompetitionOpenSinceMonth (323,348)
- CompetitionOpenSinceYear (323,348)
- Promo2SinceWeek (508,031)
- Promo2SinceYear (508,031)
- PromoInterval (508,031)

These missing values will require preprocessing
before machine learning model development.
Appropriate imputation strategies will be
investigated during the data preparation phase.

## Duplicate Values

```python
df.duplicated().sum()
```

## Observation

No duplicate records were identified within the dataset.

This suggests a relatively clean dataset and reduces
the need for duplicate removal operations during
the preprocessing stage.

# Visualization

## Sales Distribution Analysis

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Sales"],
    bins=50,
    color="orange"
)

plt.title("Sales Distribution Analysis")

plt.xlabel("Sales")

plt.ylabel("Frequency")

plt.savefig(
    "../Figures/Sales_Distributiom_Analysis.jpeg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
