# Missing Value Treatment

```python
missing_values = (
    df.isnull().sum().sort_values(ascending=False)
)

missing_values[
    missing_values>0
]
```

### Observation

Competition and promotion-related variables
contain substantial amounts of missing data.

Appropriate imputation strategies must be applied
before machine learning models are developed.

# Missing Value Strategy

```python
df["CompetitionDistance"] = (
    df["CompetitionDistance"].fillna(
        df["CompetitionDistance"].median()
    )
)

df["CompetitionOpenSinceMonth"] = (
    df["CompetitionOpenSinceMonth"].fillna(
        df["CompetitionOpenSinceMonth"].median()
    )
)

df["CompetitionOpenSinceYear"]= (
    df["CompetitionOpenSinceYear"].fillna(
        df["CompetitionOpenSinceYear"].median()
    )
)

df["Promo2SinceWeek"] = (
    df["Promo2SinceWeek"].fillna(
        df["Promo2SinceWeek"].median()
    )
)

df["Promo2SinceYear"] = (
    df["Promo2SinceYear"].fillna(
        df["Promo2SinceYear"].median()
    )
)

df["PromoInterval"] = (
    df["PromoInterval"].fillna("None")
)
```

# Verifying Missing Values

```python
df.isnull().sum()
```

### Observation

Missing values were identified in competition-related and promotion-related variables.

Median input was applied to CompetitionDistance, while zero-value and placeholder-value imputation strategies were applied to remaining cariables.

According to the treatment, no missing values were remaining among our dataset. Which made our dataset a valuable and suitable one for our future engineering and machine-learning model development.

# Date Feature Engineering

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Quarter"] = df["Date"].dt.quarter
```

# Categorical Encoding

```python
for col in [
    "StateHoliday",
    "StoreType",
    "Assortment",
    "PromoInterval"
]:
    print("\n", col)
    print(df[col].unique())
```

```python
le = LabelEncoder()
categorical_cols = [
    "StateHoliday",
    "StoreType",
    "Assortment",
    "PromoInterval"
]

for col in categorical_cols:
    df[col] = le.fit_transform(
        df[col].astype(str)
    )
```
